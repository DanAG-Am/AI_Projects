## Introducción

El procesamiento de lenguaje natural (PLN) reúne métodos computacionales para representar, analizar y extraer información del lenguaje humano. En el aprendizaje supervisado, el texto debe transformarse en características numéricas; por ello, las decisiones de preprocesamiento y representación pueden influir en qué información lingüística conserva el modelo (Manning et al., 2008).

En este reporte se aborda el análisis de sentimientos de mensajes de la plataforma X, con las clases positiva, negativa y neutral. El análisis de sentimientos busca identificar la polaridad o la opinión expresada en textos, pero presenta retos como la ambigüedad, el contexto, la negación, la ironía y el lenguaje informal. Por esta razón, no se plantea eliminar indiscriminadamente todo elemento que no sea una palabra: signos, emojis, hashtags y ciertas palabras funcionales pueden aportar señales relevantes para la tarea.

El trabajo propone un pipeline de preprocesamiento para textos crudos que contempla la inspección de la codificación y la decodificación, la normalización de mayúsculas y minúsculas, el tratamiento de URLs, menciones, hashtags, emojis, puntuación y caracteres especiales mediante reglas y expresiones regulares, y la tokenización. También se analizará el efecto de eliminar palabras cerradas de manera selectiva (conservando, por ejemplo, negaciones), incorporar etiquetas gramaticales (PoS) y comparar lematización con stemming. El objetivo es conservar un vocabulario amplio e informativo, no únicamente maximizar la cantidad de términos: la normalización debe evitar que variantes irrelevantes fragmenten las características, sin borrar matices que ayuden a distinguir sentimientos. Por ello, se estarán realizando 20+ experimentos con el fin de encontrar la combinación óptima de técnicas que maximicen la retención de información relevante para el análisis de sentimientos a través de una evaluación de las métricas de desempeño del modelo (roc-auc, f1, precisión, recall, exactitud, etc.)

Para enriquecer la representación de expresiones informales se propone explorar recursos léxicos especializados. WordNet organiza relaciones entre palabras y conceptos; SentiWordNet asigna puntuaciones de positividad, negatividad y objetividad a sentidos léxicos, y NRC EmoLex relaciona términos con emociones y polaridad. Asimismo, se considera transformar emojis y otras señales propias de redes sociales en etiquetas o características de sentimiento; VADER es un recurso diseñado para analizar texto de redes sociales y contempla reglas para señales como puntuación y uso de mayúsculas. Estas transformaciones se evaluarán como alternativas hasta que la fase experimental revele su efectividad.

El análisis incluirá diez visualizaciones distintas para estudiar la riqueza y distribución del vocabulario, como frecuencias de términos, longitud de mensajes, diversidad léxica y distribución de rangos conforme a la ley de Zipf.

Finalmente, se discutirán las ventajas y limitaciones de cada representación y las conclusiones desde el punto de vista del manejo del lenguaje.

## Metodología y enfoque

La metodología propuesta combina el preprocesamiento de texto, la extracción de características y la clasificación supervisada de sentimientos. Las etapas son:

1. **Preprocesamiento:** inspeccionar los archivos y su codificación; normalizar el texto; tratar URLs, menciones, hashtags, emojis, puntuación y caracteres especiales; tokenizar; y comparar la eliminación selectiva de palabras cerradas, el etiquetado gramatical (PoS), la lematización y el stemming.
2. **Representación:** construir características de texto con representaciones como Bag of Words y TF-IDF, y evaluar características léxicas o señales de sentimiento cuando estén disponibles.
3. **Modelado y evaluación:** comparar clasificadores con la misma partición de entrenamiento y prueba. Se proponen Naive Bayes, regresión logística y redes neuronales porque representan enfoques probabilísticos, lineales discriminativos y no lineales, respectivamente. No se asume de antemano que alguno será el mejor: se compararán con métricas por clase y validación cruzada cuando sea viable.

### Justificación de los algoritmos

Es una línea base probabilística sencilla y rápida para comparar representaciones de palabras. - **Regresión lineal y logística:** Se menciona en IBM (2025) como un algoritmo utilizado en clasificación de sentimientos. Aunque es más común en problemas de predicción continua, puede servir como referencia para comparar enfoques lineales frente a modelos más complejos. Por su parte, la regresión logística es más adecuada para clasificación, ya que modela la probabilidad de cada clase mediante la función sigmoide y permite interpretar los coeficientes de las características.
- **Redes neuronales:** El aprendizaje profundo utiliza redes neuronales, como explica IBM (2025). - **Combinación de modelos:** Se puede considerar la combinación de los tres enfoques anteriores mediante técnicas de ensamblado, como votación mayoritaria o ponderada. Esto puede mejorar la robustez y el desempeño general al aprovechar las fortalezas de cada modelo individual (IBM, 2025).

Ya que contamos con un desbalance de clase, se prioriza la métrica F1, pues proporciona un equilibrio entre precisión y recall y es más representativa del desempeño general en situaciones de desbalance.

**Preprocesamiento común**

La función `preprocess_default_tweet` aplica a todos los modelos la limpieza textual compartida: decodifica entidades HTML, normaliza URLs y menciones con tokens genéricos, conserva el contenido de los hashtags, convierte emojis y emoticonos conocidos en marcadores, pasa el texto a minúsculas, expande contracciones y tokeniza. Las negaciones se conservan; por ejemplo, `don't` se transforma en `do not`.

Las siguientes acciones no se aplican universalmente en esta función, porque son variantes que se comparan en los experimentos: eliminación de stopwords, stemming, lematización y unión de una negación con la palabra siguiente. Aplicarlas también en la limpieza base cambiaría el significado de esas comparaciones.

Dentro del preprocesamiento común, se incluyen las siguientes acciones:
- **Barajeo:** se realiza al preparar los datos, después de leerlos, con semilla 42 y manteniendo alineados textos, etiquetas y grupos. Train y test se barajan por separado; no se mezclan entre sí.
- **Armonización de etiquetas:** se realiza al cargar los registros: `objective` y `objective-or-neutral` se asignan a `neutral` para este experimento.

Así, cada modelo comparte la misma normalización básica y las variantes PLN siguen siendo comparables entre sí a nivel texto procesado.

```python
import html
import re
from collections import Counter
from sklearn.utils import shuffle

'''
Limpieza textual común para todos los modelos: normalización, expansión de
contracciones y tokenización. Las demás variantes de PLN se configuran dentro
del pipeline para poder compararlas experimentalmente.
'''
SENTIMENT_LABELS = ["negative", "neutral", "positive"]
DATA_SHUFFLE_SEED = 42
DEFAULT_PREPROCESSING_VERSION = "default_sentiment_v2" # Versión por defecto del preprocesamiento de tweets
emoji_markers = {"😊": " emoji_positive ", "😀": " emoji_positive ", "😍": " emoji_positive ", "❤": " emoji_positive ", "❤️": " emoji_positive ", "👍": " emoji_positive ", "😢": " emoji_negative ", "😭": " emoji_negative ", "😡": " emoji_negative ", "👎": " emoji_negative ", "😞": " emoji_negative "}
DEFAULT_URL_PATTERN = re.compile(r"(?:https?://|www\.)\S+", re.IGNORECASE) # Patrón para detectar URLs en el texto
DEFAULT_MENTION_PATTERN = re.compile(r"(?<![A-Za-z0-9_])@[A-Za-z0-9_]+") # Patrón para detectar menciones de usuario en el texto
DEFAULT_EMOTICONS = {":)": "emoji_positive", ":-)": "emoji_positive", ";)": "emoji_positive", ";-)": "emoji_positive", ":D": "emoji_positive", ":-D": "emoji_positive", "=D": "emoji_positive", "<3": "emoji_positive", ":(": "emoji_negative", ":-(": "emoji_negative", ";(": "emoji_negative", ":/": "emoji_neutral", ":\\": "emoji_neutral"}

def preprocess_default_tweet(text):
    '''
    Preprocesa un tweet aplicando la limpieza textual común: decodifica entidades HTML,
    normaliza URLs y menciones con tokens genéricos, conserva el contenido de los hashtags,
    convierte emojis y emoticonos conocidos en marcadores, pasa el texto a minúsculas,
    expande contracciones y tokeniza.
    Args:
        str : El texto del tweet a preprocesar.
    Returns:
        str: El texto preprocesado.
    '''
    text = html.unescape(str(text))
    text = DEFAULT_URL_PATTERN.sub(" urltoken ", text)
    text = DEFAULT_MENTION_PATTERN.sub(" usertoken ", text)
    text = re.sub(r"#([A-Za-z0-9_]+)", r" \1 ", text)
    for emoji in sorted(emoji_markers, key=len, reverse=True):
        text = text.replace(emoji, emoji_markers[emoji])
    for emoticon in sorted(DEFAULT_EMOTICONS, key=len, reverse=True):
        text = text.replace(emoticon, f" {DEFAULT_EMOTICONS[emoticon]} ")
    text = text.casefold()
    text = re.sub(r"\bwon't\b", "will not", text)
    text = re.sub(r"\bcan't\b", "can not", text)
    text = re.sub(r"\b([a-z]+)n't\b", r"\1 not", text)
    return " ".join(re.findall(r"(?u)\b\w+\b", text))
```

**Manejo de duplicados entre training y test**

Se asegura que no haya registros idénticos en los conjuntos de entrenamiento y prueba. Esto evita que el modelo tenga acceso a ejemplos del test durante el entrenamiento, lo que podría inflar artificialmente las métricas de desempeño. Esto se identificó comparando los textos de ambos conjuntos y eliminando los duplicados encontrados. Adicionalmente, se veía en el desempeño de los modelos, donde el rendimiento mejoraba de manera inesperada para ciertos k-folds. 

```python
from collections import defaultdict
from pathlib import Path
import pandas as pd

def load_duplicate_demo_records(path, split_name):
    '''
    Carga los registros duplicados de un archivo de datos.
    Args:
        path (Path): La ruta al archivo de datos.
        split_name (str): El nombre del conjunto de datos (por ejemplo, "train" o "test").
    Returns:
        list: Una lista de diccionarios que representan los registros duplicados.
    '''
    rows = []
    with open(path, encoding="utf-8") as dataset_file:
        for line in dataset_file:
            columns = line.rstrip("\r\n").split("\t", maxsplit=3)
            if len(columns) != 4:
                continue
            label = columns[2].strip().casefold()
            if label in {"objective", "objective-or-neutral"}:
                label = "neutral"
            text = columns[3].strip()
            if label in SENTIMENT_LABELS and text:
                rows.append({"tweet_id": columns[0].strip(), "text": text, "label": label, "split": split_name})
    return rows

duplicate_demo_records = load_duplicate_demo_records(Path("originallTraining.txt"), "train") + load_duplicate_demo_records(Path("originalTest.txt"), "test")
duplicate_demo_parents = list(range(len(duplicate_demo_records)))

def duplicate_demo_find(index):
    '''
    Encuentra el grupo representativo de un registro duplicado, para ello, unimos por rango los índices de los registros duplicados.
    Args:
        index (int): El índice del registro duplicado.
    Returns:
        int: El índice del grupo representativo al que pertenece el registro duplicado.
    '''
    while duplicate_demo_parents[index] != index:
        duplicate_demo_parents[index] = duplicate_demo_parents[duplicate_demo_parents[index]]
        index = duplicate_demo_parents[index]
    return index

duplicate_demo_seen = {}
for index, record in enumerate(duplicate_demo_records):
    normalized_text = " ".join(record["text"].casefold().split())
    keys = [f"text:{normalized_text}"]
    if record["tweet_id"]:
        keys.append(f"id:{record['tweet_id']}")
    for key in keys:
        if key in duplicate_demo_seen:
            duplicate_demo_parents[duplicate_demo_find(index)] = duplicate_demo_find(duplicate_demo_seen[key])
        else:
            duplicate_demo_seen[key] = index

duplicate_demo_group_ids = [duplicate_demo_find(index) for index in range(len(duplicate_demo_records))]
duplicate_demo_members = defaultdict(list)
for record, group_id in zip(duplicate_demo_records, duplicate_demo_group_ids):
    duplicate_demo_members[group_id].append(record)
duplicate_demo_repeated_groups = {group_id: members for group_id, members in duplicate_demo_members.items() if len(members) > 1}
duplicate_demo_cross_split_groups = {group_id: members for group_id, members in duplicate_demo_repeated_groups.items() if len({record["split"] for record in members}) > 1}
if len(duplicate_demo_repeated_groups) == 0:
    print("Sin duplicados entre training y testing")
else:
    print(f"Registros leídos: {len(duplicate_demo_records)}")
    print(f"Grupos con duplicados (mismo ID o texto normalizado): {len(duplicate_demo_repeated_groups)}")
    print(f"Registros repetidos agrupados: {sum(len(members) - 1 for members in duplicate_demo_repeated_groups.values())}")
    print(f"Grupos duplicados que aparecen tanto en train como en test: {len(duplicate_demo_cross_split_groups)}")
    duplicate_demo_examples = [{"group_id": group_id, "copies": len(members), "splits": ", ".join(sorted({record["split"] for record in members})), "tweet_id": members[0]["tweet_id"], "text": members[0]["text"][:100]} for group_id, members in list(duplicate_demo_repeated_groups.items())[:5]]
    display(pd.DataFrame(duplicate_demo_examples))
```

## Creación de los top 20 modelos

El flujo principal conserva seis bloques de código para crear los checkpoints:

1. Limpieza común: UTF-8, entidades HTML, URLs, menciones, hashtags, emojis, minúsculas y negaciones.
2. Pipeline configurable: variaciones PLN, representación vectorial y clasificador. Naive Bayes utiliza exclusivamente MultinomialNB.
3. Preparación de datos: armonización de etiquetas, filtrado de duplicados y grupos contradictorios, barajado reproducible y tres folds agrupados por ID/texto.
4. Evaluación de 48 candidatos iniciales.
5. Incorporación de ocho variantes controladas adicionales, para un total de 56 configuraciones únicas.
6. Selección por macro-F1 de CV, entrenamiento y guardado de 20 pipelines completos en `saved_models_pt`, numerados por ranking.

El vocabulario se aprende dentro de cada fold. Prueba no interviene en la selección; se usa para comprobar que los checkpoints recargados reproduzcan sus predicciones. La recarga requiere las definiciones del preprocesador y del clasificador personalizado incluidas aquí. Solo deben cargarse checkpoints de confianza.

El benchmark de RoBERTa preentrenado se ejecuta por separado del top 20 y guarda su modelo y tokenizador en `saved_roberta_sentiment`. No modifica el ranking ni los checkpoints locales. Las demos y visualizaciones retiradas no intervienen en ninguno de estos dos flujos; esta versión no incluye todos los modelos probados, solo aquellos que se consideraron más relevantes para el top 20.

En el caso de WordNet, se trabaja con synsets, que son conjuntos de sinónimos que representan un concepto específico. Esto permite evaluar el sentimiento de un token considerando su contexto semántico. Sin embargo, esta aproximación tiene limitaciones por la ambigüedad y la variabilidad del lenguaje. (GeeksforGeeks, 2022)

```python
from functools import lru_cache # -> para guardar los resultados de funciones costosas en caché, mejorando el rendimiento al evitar cálculos repetidos: https://www.alg0.dev/es/lru-cache/
from pathlib import Path
import nltk
import numpy as np
import pandas as pd
from nltk.corpus import sentiwordnet, wordnet # -> taxonomia lexica especializada
from nltk.stem import SnowballStemmer, WordNetLemmatizer # -> para el stemming y lematización de palabras
from scipy.special import softmax # -> RNN
from sklearn.base import BaseEstimator, ClassifierMixin, TransformerMixin # -> clases base para estimadores, clasificadores y transformadores de scikit-learn
from sklearn.feature_extraction.text import CountVectorizer, ENGLISH_STOP_WORDS, TfidfVectorizer # -> para la extracción de características de texto
from sklearn.linear_model import LinearRegression, LogisticRegression # -> modelos lineales de regresión y clasificación
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_auc_score # -> métricas de evaluación de modelos
from sklearn.naive_bayes import MultinomialNB # -> clasificador Naive Bayes multinomial
from sklearn.neural_network import MLPClassifier # -> clasificador de red neuronal multicapa
from sklearn.pipeline import FeatureUnion, Pipeline # -> para crear pipelines de procesamiento y combinación de características
from sklearn.preprocessing import label_binarize # -> para convertir etiquetas en formato binario (one hot)

NEGATIONS = {"no", "nor", "not", "never", "neither", "n't"}
EXPERIMENT_STOP_WORDS = set(ENGLISH_STOP_WORDS) - NEGATIONS # -> stop words experimentales, excluyendo las negaciones
experiment_stemmer = SnowballStemmer("english")
experiment_lemmatizer = WordNetLemmatizer()

def wordnet_pos(tag):
    '''
    Convierte una etiqueta de parte del habla de NLTK a la correspondiente etiqueta de WordNet.
    Parámetros:
    tag (str): Etiqueta de parte del habla de NLTK.
    Retorna:
    str: Etiqueta de parte del habla de WordNet.
    '''
    return {"J": "a", "V": "v", "R": "r"}.get(tag[:1], "n")

@lru_cache(maxsize=30000) # -> guarda en caché los resultados de la función para mejorar el rendimiento, hasta un máximo de 30000 entradas
def sentiment_marker(token, pos):
    '''
    Marca el sentimiento de un token dado su parte del habla utilizando SentiWordNet.
    Parámetros:
    token (str): Token a evaluar.
    pos (str): Etiqueta de parte del habla de WordNet.
    Retorna:
    str o None: "lexicon_positive" si el token tiene sentimiento positivo,
                "lexicon_negative" si el token tiene sentimiento negativo,
                None como salida del bucle de evaluación del sentimiento -> limitación de SentiWordNet.
    '''
    synsets = wordnet.synsets(token, pos=pos)
    if len(synsets) == 0:
        return None
    scores = sentiwordnet.senti_synset(synsets[0].name())
    if scores.pos_score() > scores.neg_score():
        return "lexicon_positive"
    if scores.neg_score() > scores.pos_score():
        return "lexicon_negative"
    return None

class TweetPreprocessor(TransformerMixin, BaseEstimator):
    def __init__(self, mode="base"):
        self.mode = mode

    def fit(self, texts, labels=None):
        return self

    def transform(self, texts):
        '''
        Transforma los textos originales en una representación procesada según el modo especificado.
        Parámetros:
        texts (list of str): Lista de textos originales a procesar.
        Retorna:
        list of str: Lista de textos transformados según el modo especificado.
        Modos:
        - "base": Retorna el texto original sin modificaciones.
        - "stop": Elimina las stop words experimentales, excluyendo las negaciones.
        - "stem": Aplica stemming a los tokens, excepto a los emojis.
        - "lemma": Aplica lematización a los tokens según su parte del habla.
        - "pos": Agrega etiquetas de parte del habla a los tokens.
        - "sentiwordnet": Marca el sentimiento de los tokens utilizando SentiWordNet.
        - "lexicon": Marca el sentimiento de los tokens utilizando un lexicon predefinido.
        - "negation": Marca las negaciones en los tokens.
        - "social_signals": Agrega señales sociales como mayúsculas, signos de exclamación, signos de interrogación y elongaciones.
        '''
        transformed = []
        for original_text in texts:
            text = preprocess_default_tweet(original_text)
            tokens = text.split()
            if self.mode == "stop":
                tokens = [token for token in tokens if token not in EXPERIMENT_STOP_WORDS]
            elif self.mode == "stem":
                tokens = [token if token.startswith("emoji_") else experiment_stemmer.stem(token) for token in tokens]
            elif self.mode in {"lemma", "pos", "sentiwordnet"}:
                tagged = nltk.pos_tag(tokens, lang="eng")
                if self.mode == "lemma":
                    tokens = [experiment_lemmatizer.lemmatize(token, wordnet_pos(tag)) for token, tag in tagged]
                elif self.mode == "pos":
                    tokens += [f"pos_{tag.casefold()}" for token, tag in tagged]
                else:
                    for token, tag in tagged:
                        marker = sentiment_marker(token, wordnet_pos(tag))
                        if marker is not None:
                            tokens.append(marker)
            elif self.mode == "lexicon":
                if {"good", "great", "love", "happy", "amazing"}.intersection(tokens):
                    tokens.append("lexicon_positive")
                if {"bad", "awful", "hate", "sad", "terrible"}.intersection(tokens):
                    tokens.append("lexicon_negative")
            elif self.mode == "negation":
                tokens = re.sub(r"\b(?:not|never|no|nor)\s+(\w+)\b", r"not_\1", text).split()
            elif self.mode == "social_signals":
                if re.search(r"\b[A-Z]{2,}\b", original_text):
                    tokens.append("allcaps_signal")
                if "!" in original_text:
                    tokens.append("exclamation_signal")
                if "?" in original_text:
                    tokens.append("question_signal")
                if re.search(r"(.)\1{2,}", original_text, re.IGNORECASE):
                    tokens.append("elongation_signal")
            transformed.append(" ".join(tokens))
        return transformed

class OneHotLinearRegressionClassifier(ClassifierMixin, BaseEstimator):
    def fit(self, features, labels):
        self.classes_ = np.unique(labels)
        targets = (np.asarray(labels)[:, None] == self.classes_[None, :]).astype(float)
        self.regressor_ = LinearRegression().fit(features, targets)
        return self

    def predict_proba(self, features):
        return softmax(self.regressor_.predict(features), axis=1)

    def predict(self, features):
        return self.classes_[self.predict_proba(features).argmax(axis=1)]

def make_text_pipeline(config):
    '''
    pipeline de procesamiento de texto según la configuración especificada.
    Parámetros:
    config (dict): Diccionario de configuración que especifica el vectorizador, el algoritmo y sus parámetros.
    Retorna:
    Pipeline: Pipeline de procesamiento de texto y clasificación según la configuración.
    Parametros de config:
    - "vectorizer" (str): Tipo de vectorizador ("BoW", "One-hot", "TF-IDF", "Char TF-IDF", "Word + char TF-IDF").
    - "max_features" (int, opcional): Número máximo de características a considerar.
    - "min_df" (int, opcional): Frecuencia mínima de documentos para incluir una característica.
    - "ngram_range" (tuple, opcional): Rango de n-gramas a considerar.
    - "sublinear_tf" (bool, opcional): Si se debe aplicar TF sublineal.
    - "binary" (bool, opcional): Si se debe usar representación binaria para One-hot.
    - "mode" (str, opcional): Modo de preprocesamiento de los tweets ("base", "stop", "stem", "lemma", "pos", "sentiwordnet", "lexicon", "negation", "social_signals").
    '''
    # inicialización de parámetros de configuración
    representation = config["vectorizer"]
    max_features = config.get("max_features", 5000)
    min_df = config.get("min_df", 1)
    ngram_range = config.get("ngram_range", (1, 1))
    sublinear_tf = config.get("sublinear_tf", False)

    if representation == "BoW" or representation == "One-hot":
        binary = representation == "One-hot" or config.get("binary", False)
        features = CountVectorizer(lowercase=False, max_features=max_features, min_df=min_df, ngram_range=ngram_range, binary=binary)
    elif representation == "TF-IDF":
        features = TfidfVectorizer(lowercase=False, max_features=max_features, min_df=min_df, ngram_range=ngram_range, sublinear_tf=sublinear_tf)
    elif representation == "Char TF-IDF":
        features = TfidfVectorizer(lowercase=False, max_features=max_features, min_df=min_df, analyzer="char", ngram_range=ngram_range, sublinear_tf=sublinear_tf)
    elif representation == "Word + char TF-IDF":
        word_features = TfidfVectorizer(lowercase=False, max_features=max_features, min_df=min_df, ngram_range=ngram_range, sublinear_tf=sublinear_tf)
        char_features = TfidfVectorizer(lowercase=False, max_features=max_features, min_df=min_df, analyzer="char", ngram_range=(3, 5), sublinear_tf=sublinear_tf)
        features = FeatureUnion([("word", word_features), ("char", char_features)])
    else:
        print("Representación desconocida; se usará BoW.") # -> para poder usar el bucle
        features = CountVectorizer(lowercase=False, max_features=max_features, min_df=min_df, ngram_range=ngram_range)

    algorithm = config["algorithm"]
    if algorithm == "MultinomialNB":
        classifier = MultinomialNB(alpha=config.get("alpha", 1.0))
    elif algorithm == "LogisticRegression":
        classifier = LogisticRegression(C=config.get("C", 1.0), class_weight=config.get("class_weight"), max_iter=1800, random_state=DATA_SHUFFLE_SEED)
    elif algorithm == "MLPClassifier":
        classifier = MLPClassifier(hidden_layer_sizes=config.get("hidden_layer_sizes", (64,)), alpha=config.get("alpha", 0.01), max_iter=80, early_stopping=True, n_iter_no_change=8, random_state=DATA_SHUFFLE_SEED)
    elif algorithm == "LinearRegression":
        classifier = OneHotLinearRegressionClassifier()
    else:
        print("Clasificador desconocido; se usará MultinomialNB.") # -> Para poder usar el bucle
        classifier = MultinomialNB(alpha=config.get("alpha", 1.0))

    preprocessing = TweetPreprocessor(config.get("preprocessing", "base"))
    return Pipeline([("preprocessing", preprocessing), ("features", features), ("classifier", classifier)])

def calculate_metrics(true_labels, predicted_labels, probabilities=None):
    '''
    Calcula métricas de evaluación para un conjunto de etiquetas verdaderas y predichas.
    Parámetros:
    true_labels (list): Lista de etiquetas verdaderas.
    predicted_labels (list): Lista de etiquetas predichas.
    probabilities (ndarray, opcional): Probabilidades predichas para cada clase.
    Retorna:
    dict: Diccionario con las métricas de evaluación (accuracy, precision_macro, recall_macro, f1_macro, roc_auc_macro).
    '''
    metrics = {"accuracy": accuracy_score(true_labels, predicted_labels), "precision_macro": precision_score(true_labels, predicted_labels, labels=SENTIMENT_LABELS, average="macro", zero_division=0), "recall_macro": recall_score(true_labels, predicted_labels, labels=SENTIMENT_LABELS, average="macro", zero_division=0), "f1_macro": f1_score(true_labels, predicted_labels, labels=SENTIMENT_LABELS, average="macro", zero_division=0)}
    if probabilities is not None and set(true_labels) == set(SENTIMENT_LABELS): # -> para cuando roc auc sea pertinente
        metrics["roc_auc_macro"] = roc_auc_score(label_binarize(true_labels, classes=SENTIMENT_LABELS), probabilities, average="macro")
    else:
        metrics["roc_auc_macro"] = np.nan
    return metrics

def evaluate_text_pipeline(model, texts, labels):
    '''
    Evalúa un modelo de procesamiento de texto en un conjunto de datos.
    Parámetros:
    model (Pipeline): Modelo de procesamiento de texto.
    texts (list): Lista de textos a evaluar.
    labels (list): Lista de etiquetas verdaderas correspondientes a los textos.
    Retorna:
    dict: Diccionario con las métricas de evaluación calculadas.
    '''
    predictions = model.predict(texts)
    probabilities = None
    if hasattr(model, "predict_proba"):
        class_order = list(model.classes_)
        probabilities = model.predict_proba(texts)[:, [class_order.index(label) for label in SENTIMENT_LABELS]]
    return calculate_metrics(labels, predictions, probabilities)
```

## Producción de los 20 mejores modelos

Se seleccionarán y evaluarán los 20 mejores modelos según las métricas de validación cruzada. Esto permitirá identificar los modelos más prometedores tomando en cuenta su capacidad para generalizar a nuevos datos en varias ocasiones.

```python
from collections import defaultdict
from sklearn.model_selection import StratifiedGroupKFold

def read_experiment_records(path):
    '''
    Lee los registros de un experimento desde un archivo.
    Parámetros:
    path (str): Ruta al archivo que contiene los registros del experimento.
    Retorna:
    list: Lista de diccionarios con los registros del experimento.
    '''
    records = []
    with open(path, encoding="utf-8") as dataset_file:
        for line in dataset_file:
            columns = line.rstrip("\r\n").split("\t", maxsplit=3)
            if len(columns) != 4:
                continue
            label = columns[2].strip().casefold()
            if label == "objective" or label == "objective-or-neutral":
                label = "neutral"
            if label in SENTIMENT_LABELS and columns[3].strip():
                records.append({"tweet_id": columns[0].strip(), "text": columns[3].strip(), "label": label})
    return records

def duplicate_groups(records):
    '''
    Identifica grupos de registros duplicados en un conjunto de datos.
    Parámetros:
    records (list): Lista de diccionarios con los registros del experimento.
    Retorna:
    list: Lista de identificadores de grupo para cada registro.
    '''
    parents = list(range(len(records)))
    def find(index):
        while parents[index] != index:
            parents[index] = parents[parents[index]]
            index = parents[index]
        return index
    seen = {}
    for index, record in enumerate(records):
        normalized_text = " ".join(record["text"].casefold().split())
        keys = ["text:" + normalized_text]
        if record["tweet_id"]:
            keys.append("id:" + record["tweet_id"])
        for key in keys:
            if key in seen:
                first_group = find(index)
                second_group = find(seen[key])
                parents[first_group] = second_group
            else:
                seen[key] = index
    return [find(index) for index in range(len(records))]

train_records = read_experiment_records("originallTraining.txt")
test_records = read_experiment_records("originalTest.txt")
records = train_records + test_records
groups = duplicate_groups(records)
training_group_ids = set(groups[:len(train_records)])
group_labels = defaultdict(set)
for record, group in zip(records, groups):
    group_labels[group].add(record["label"])
conflicting_groups = {group for group, labels in group_labels.items() if len(labels) > 1}
train_indices = [index for index in range(len(train_records)) if groups[index] not in conflicting_groups]
test_indices = []
seen_test_groups = set()
for index in range(len(train_records), len(records)):
    group = groups[index]
    if group not in training_group_ids and group not in conflicting_groups and group not in seen_test_groups:
        test_indices.append(index)
        seen_test_groups.add(group)

experiment_train_texts, experiment_train_labels, experiment_train_groups = shuffle(
    [records[index]["text"] for index in train_indices],
    [records[index]["label"] for index in train_indices],
    [groups[index] for index in train_indices],
    random_state=DATA_SHUFFLE_SEED,
)
experiment_test_texts, experiment_test_labels, experiment_test_groups = shuffle(
    [records[index]["text"] for index in test_indices],
    [records[index]["label"] for index in test_indices],
    [groups[index] for index in test_indices],
    random_state=DATA_SHUFFLE_SEED,
)
if set(experiment_train_groups).isdisjoint(experiment_test_groups):
    print("Train y test no comparten grupos duplicados.")
else:
    print("Aviso: quedan grupos compartidos entre train y test.")

experiment_cv = StratifiedGroupKFold(n_splits=3, shuffle=True, random_state=DATA_SHUFFLE_SEED)
experiment_splits = list(experiment_cv.split(experiment_train_texts, experiment_train_labels, groups=experiment_train_groups))

def run_experiments(configurations):
    '''
    Ejecuta experimentos de validación cruzada para un conjunto de configuraciones de modelos.
    Parámetros:
    configurations (list): Lista de diccionarios con las configuraciones de los modelos a evaluar.
    Retorna:
    pd.DataFrame: DataFrame con los resultados de validación cruzada y las características de cada modelo.
    '''
    rows = []
    for config in configurations:
        fold_metrics = []
        feature_counts = []
        for train_indices, validation_indices in experiment_splits:
            model = make_text_pipeline(config)
            model.fit(
                [experiment_train_texts[index] for index in train_indices],
                [experiment_train_labels[index] for index in train_indices],
            )
            validation_texts = [experiment_train_texts[index] for index in validation_indices]
            validation_labels = [experiment_train_labels[index] for index in validation_indices]
            fold_metrics.append(evaluate_text_pipeline(model, validation_texts, validation_labels))
            feature_counts.append(len(model.named_steps["features"].get_feature_names_out()))
        row = {"model_id": config["model_id"], "preprocessing": config.get("preprocessing", "base"), "vectorizer": config["vectorizer"], "algorithm": config["algorithm"], "justification": config["purpose"], "config": config, "vocabulary_size_cv": float(np.mean(feature_counts))}
        for metric in fold_metrics[0]:
            values = [result[metric] for result in fold_metrics]
            row[metric + "_cv_mean"] = float(np.mean(values))
            row[metric + "_cv_std"] = float(np.std(values, ddof=1))
        rows.append(row)
    results = pd.DataFrame(rows)
    return results.sort_values(["f1_macro_cv_mean", "model_id"], ascending=[False, True]).reset_index(drop=True)
```

## Discusión de resultados

### Protocolo de comparación

Se evaluaron **56 configuraciones locales** con el mismo pipeline de limpieza, representación y clasificación. Los archivos se leen como UTF-8; las etiquetas `objective` y `objective-or-neutral` se asignan a `neutral`. Antes del filtrado se cuentan 6,165 registros de entrenamiento y 974 de prueba. Después de excluir grupos contradictorios, solapamientos y duplicados de prueba, se utilizan **6,136 registros de entrenamiento y 966 de prueba**.

Los datos se barajan con semilla **42**, manteniendo texto, etiqueta y grupo alineados. La selección emplea **StratifiedGroupKFold de tres folds**, agrupados por ID o texto duplicado. Cada vectorizador aprende el vocabulario y, cuando corresponde, el IDF exclusivamente sobre el entrenamiento de su fold. El top 20 se ordena por la media de **macro-F1 de CV**, no por prueba. 

La limpieza base conserva señales relevantes: las URLs y menciones se sustituyen por tokens genéricos; los hashtags conservan su contenido; emojis y emoticonos se convierten en marcadores de sentimiento; las contracciones se expanden sin borrar negaciones. Las variantes de PoS, morfología, léxicos y señales sociales se aplican dentro del pipeline. Las señales de mayúsculas y puntuación se extraen del texto original, antes de retirarlas.

### Justificación de las variaciones

Las variaciones contrastan hipótesis lingüísticas y de aprendizaje, no únicamente valores arbitrarios de hiperparámetros:

- **Frecuencia frente a presencia binaria:** BoW conserva repeticiones; one-hot por palabra produce un vector multi-hot por tweet y evita que una palabra repetida domine la representación. 
- **Unigramas, bigramas y trigramas:** aumentar el orden conserva frases locales, pero introduce más dimensiones y expresiones poco frecuentes. Por ello se combina con límites de vocabulario y `min_df`, ya que se observó que sin estos parametros, el rendimiento degradaba entre todas las combinaciones similares, dejando un escenario injusto.
- **BoW frente a TF-IDF:** TF-IDF reduce el peso de palabras frecuentes en muchos documentos; su utilidad depende del clasificador. Al ser una ponderación más elaborada, se utiliza para evaluar si la mejora en la representación del texto se traduce en un mejor rendimiento del modelo.
- **Palabras cerradas y negaciones:** eliminar stopwords puede reducir ruido, pero conservar `not`, `no`, `never` y otras negaciones evita perder cambios de polaridad. Unir negación y palabra siguiente es otra hipótesis, con el riesgo de fragmentar el vocabulario en características raras.
- **Stemming y lematización:** buscan reunir variantes morfológicas. El stemming es más agresivo; la lematización usa PoS y WordNet, pero depende de la calidad del etiquetado y puede perder matices.
- **PoS y taxonomías léxicas:** las etiquetas gramaticales añaden señales sintácticas; SentiWordNet añade marcadores positivos o negativos basados en sentidos compatibles con PoS. 
- **Caracteres y representación mixta:** los n-gramas de caracteres toleran errores de escritura y formas informales; combinarlos con palabras prueba si aportan información complementaria, a costa de memoria y menor interpretabilidad.
- **Clasificadores y regularización:** MultinomialNB ofrece una referencia probabilística rápida; regresión logística aprende pesos discriminativos. La regresión lineal sobre objetivos one-hot se incluye como referencia, no como un clasificador probabilístico.
- **Balanceo de clases:** `class_weight='balanced'` en regresión logística compensa la menor representación de negativos. Macro-F1 asigna la misma importancia a cada clase, por lo que no basta con mejorar la exactitud de la clase mayoritaria.

Esta conclusión describe las configuraciones probadas, más no se debe interpretar como una inferioridad general de las redes neuronales ante la clasificación de textos, ya que como vemos con roBERTa preentrenado, estas arquitecturas pueden alcanzar un rendimiento superior sin necesidad de ajustes locales.

### Pruebas controladas del efecto de PLN - Maximizando la configuración de regresión logística.

Una vez se encuentra que las configuraciones de reg log son las que están predominando en el top 20, se mantuvieron constantes **regresión logística**, `C=1.0`, `class_weight='balanced'`, `max_iter=1800`, semilla 42 y **TF-IDF de unigramas**, `max_features=5000`, `min_df=2`, `sublinear_tf=False`, normalización L2. Solo cambió el tratamiento lingüístico. La desviación entre folds refleja la variabilidad de cada configuración bajo estas condiciones, midiendo el aumento o reducción de la métrica f1 entre folds. ### Conclusiones y limitaciones

BoW ofrece una línea base; la presencia binaria descarta repeticiones y TF-IDF atenúa términos con alta frecuencia. Los n-gramas conservan contexto local, pero aumentan dimensionalidad y rareza. Los caracteres toleran variantes informales con menor interpretabilidad. El enriquecimiento léxico proporciona señales compartidas entre palabras, aunque depende de la cobertura y de los sentidos seleccionados.

La evidencia indica una mejora al incorporar SentiWordNet, stemming y señales sociales en la regresión logística balanceada. Maximizar vocabulario no equivale a maximizar desempeño: reducir fragmentación morfológica o incorporar marcadores útiles puede ser más importante que conservar todas las variantes de escritura.

Los 20 checkpoints locales se seleccionan por F1 macro y conservan el pipeline completo. La clase negativa continúa siendo difícil: en el ganador local, su F1 de prueba es aproximadamente 0.5304, frente a 0.6840 en neutral y 0.6478 en positiva. Por ello se prioriza macro-F1 sobre exactitud u otras métricas para la selección de modelos.

Dentro de las acciones que más contribuyen de manera positiva al desempeño del modelo de forma general se encuentran: balance de clases, incorporación de SentiWordNet, stemming y señales sociales.
