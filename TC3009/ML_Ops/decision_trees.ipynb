```python
# Step 1: Build the dataframe
import pandas as pd
import numpy as np

data = {
    "stream": ["false", "true", "true", "false", "false", "true", "true"],
    "slope": ["steep", "moderate", "steep", "steep", "flat", "steep", "steep"],
    "elevation": ["high", "low", "medium", "medium", "high", "highest", "high"], 
    "vegetation": ["chapparal", "riparian", "riparian", "chapparal", "conifer", "conifer", "chapparal"]
}
df = pd.DataFrame(data)
print(df)

# Step 2: Define the target variable and features for the decision tree
target = "vegetation"
features = ["stream", "slope", "elevation"]

# Step 3: Split the data into features (X) and target (y)
X = df[features].copy()
y = df[target]

# Step 4: Print the features and target to verify
print("\nFeatures (X):")
print(X)
print("\nTarget (y):")
print(y)

# Step 5: Define numbers for categorical encoding
stream_mapping = {"false": 0, "true": 1}
slope_mapping = {"flat": 0, "moderate": 1, "steep": 2}
elevation_mapping = {"low": 0, "medium": 1, "high": 2, "highest": 3}
vegetation_mapping = {"chapparal": 0, "riparian": 1, "conifer": 2}

X["stream"] = X["stream"].map(stream_mapping)
X["slope"] = X["slope"].map(slope_mapping)
X["elevation"] = X["elevation"].map(elevation_mapping)
y = y.map(vegetation_mapping)

# Step 6: Print the encoded features and target to verify
print("\nEncoded Features (X):")
print(X)
print("\nEncoded Target (y):")
print(y)

# Step 7: Manual building of the decision tree with featuresand calculate entropy

# Calculate the entropy of the target variable (y)
def entropy(y):
    counts = np.bincount(y)
    probabilities = counts / len(y)
    entropy = -np.sum([p * np.log2(p) for p in probabilities if p > 0])
    return entropy

print("\nEntropy of the target variable (y):")
print(entropy(y))

# Calculate the entropy of the target variable (y) for each value of the "stream" feature
for stream_value in X["stream"].unique():
    subset_y = y[X["stream"] == stream_value]
    subset_entropy = entropy(subset_y)
    print(f"Entropy of the target variable (y) for stream = {stream_value}: {subset_entropy}")
total_entropy = entropy(y)
stream_info_gain = total_entropy
for stream_value in X["stream"].unique():
    subset_y = y[X["stream"] == stream_value]
    subset_entropy = entropy(subset_y)
    stream_info_gain -= (len(subset_y) / len(y)) * subset_entropy
print(f"Information gain for the 'stream' feature: {stream_info_gain}")

# Calculate the entropy of the target variable (y) for each value of the "slope" feature
for slope_value in X["slope"].unique():
    subset_y = y[X["slope"] == slope_value]
    subset_entropy = entropy(subset_y)
    print(f"Entropy of the target variable (y) for slope = {slope_value}: {subset_entropy}")
total_entropy = entropy(y)
slope_info_gain = total_entropy
for slope_value in X["slope"].unique():
    subset_y = y[X["slope"] == slope_value]
    subset_entropy = entropy(subset_y)
    slope_info_gain -= (len(subset_y) / len(y)) * subset_entropy
print(f"Information gain for the 'slope' feature: {slope_info_gain}")

# Calculate the entropy of the target variable (y) for each value of the "elevation" feature
for elevation_value in X["elevation"].unique():
    subset_y = y[X["elevation"] == elevation_value]
    subset_entropy = entropy(subset_y)
    print(f"Entropy of the target variable (y) for elevation = {elevation_value}: {subset_entropy}")
total_entropy = entropy(y)
elevation_info_gain = total_entropy
for elevation_value in X["elevation"].unique():
    subset_y = y[X["elevation"] == elevation_value]
    subset_entropy = entropy(subset_y)
    elevation_info_gain -= (len(subset_y) / len(y)) * subset_entropy
print(f"Information gain for the 'elevation' feature: {elevation_info_gain}")
```

## Conclusion
Elevation is the best feature to split on first: it has the highest information gain.
It reduces uncertainty about vegetation more than slope or stream in this dataset.
