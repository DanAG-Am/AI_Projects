```python
import torch
import torch.nn as nn
```

```python
# ------------------------------------------------------------------------------
# EXERCISE 1: Tensor Creation and Datatypes
# ------------------------------------------------------------------------------
def create_basic_tensors():
    """
    1. Create a scalar 'a' with value 42.
    2. Create a 1D vector 'v' containing values [1.0, 2.0, 3.0, 4.0] with dtype float32.
    3. Cast 'v' to float16 and store it in 'v_float16'.
    """
    # TODO: Create a scalar containing value 42
    a = torch.tensor(42)

    # TODO: Create a 1D vector with values [1.0, 2.0, 3.0, 4.0] of type float32
    v = torch.tensor([1.0, 2.0, 3.0, 4.0], dtype=torch.float32)

    # TODO: Convert vector 'v' to float16
    v_float16 = v.to(torch.float16)

    return a, v, v_float16

print(create_basic_tensors())
```

```python
# ------------------------------------------------------------------------------
# EXERCISE 2: Matrix Multiplication & Shape Alignment
# ------------------------------------------------------------------------------
def align_and_multiply(tensor_a, tensor_b):
    """
    Given tensor_a of shape (3, 2) and tensor_b of shape (3, 2):
    1. Transpose tensor_b so its shape matches matrix multiplication rules with tensor_a.
    2. Perform matrix multiplication between tensor_a and the transposed tensor_b.

    Args:
        tensor_a (torch.Tensor): Tensor of shape (3, 2)
        tensor_b (torch.Tensor): Tensor of shape (3, 2)

    Returns:
        torch.Tensor: Resulting matrix multiplication product of shape (3, 3)
    """
    # TODO: Transpose tensor_b and perform matrix multiplication with tensor_a
    result = torch.matmul(tensor_a, tensor_b.T)
    return result

print(align_and_multiply(torch.tensor([[1, 2], [3, 4], [5, 6]], dtype=torch.float32),
                        torch.tensor([[7, 8], [9, 10], [11, 12]], dtype=torch.float32)))
```

```python
# ------------------------------------------------------------------------------
# EXERCISE 3: Reshaping, Squeezing, and Linear Layers
# ------------------------------------------------------------------------------
def apply_linear_transformation(x):
    """
    Given an input vector x of shape (4,):
    1. Reshape or unsqueeze x to have a batch dimension, turning shape (4,) into (1, 4).
    2. Pass the reshaped input through a torch.nn.Linear layer configured to map
       in_features=4 to out_features=2.

    Args:
        x (torch.Tensor): Input tensor of shape (4,)

    Returns:
        torch.Tensor: Transformed output tensor of shape (1, 2)
    """
    torch.manual_seed(42)  # For reproducible weight initialization

    # TODO: Instantiate a Linear layer with in_features=4 and out_features=2
    linear_layer = torch.nn.Linear(in_features=4, out_features=2)

    # TODO: Add a batch dimension to x so its shape becomes (1, 4)
    x_batched = x.unsqueeze(0)
    print(x_batched.shape)
    # TODO: Pass x_batched through the linear layer
    output = linear_layer(x_batched)
    print(output.shape)

    return output

print(apply_linear_transformation(torch.tensor([1.0, 2.0, 3.0, 4.0])))
```

```python
# ------------------------------------------------------------------------------
# EXERCISE 4 (NEW): Tensor Aggregations & Keeping Dimensions
# ------------------------------------------------------------------------------
def summarize_batch_features(features):
    """
    Given a feature matrix 'features' of shape (B, D) representing a batch of B samples
    and D feature dimensions:
    1. Calculate the mean across the batch dimension (dim=0) while keeping dimensions (keepdim=True).
    2. Calculate the variance across the batch dimension (dim=0) while keeping dimensions (keepdim=True).
    3. Normalize the input tensor using standard score formula: (features - mean) / sqrt(var + 1e-5).

    Args:
        features (torch.Tensor): Input matrix of shape (B, D)

    Returns:
        tuple[torch.Tensor, torch.Tensor]: (mean_tensor, normalized_features)
            - mean_tensor shape: (1, D)
            - normalized_features shape: (B, D)
    """
    # TODO: Calculate batch mean with keepdim=True
    mean = features.mean(dim=0, keepdim=True)

    # TODO: Calculate batch variance with keepdim=True
    var = features.var(dim=0, keepdim=True)

    # TODO: Compute standard score normalized tensor: (features - mean) / sqrt(var + 1e-5)
    normalized = (features - mean) / torch.sqrt(var + 1e-5)

    return mean, normalized
```
